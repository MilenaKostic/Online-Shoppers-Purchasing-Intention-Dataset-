# 3.Ekstremne vrednosti

```python
num_columns = ['Administrative', 'Administrative_Duration', 'Informational',
               'Informational_Duration', 'ProductRelated', 'ProductRelated_Duration',
               'BounceRates', 'ExitRates', 'PageValues', 'SpecialDay']

for col in num_columns:
    plt.figure(figsize=(10,4))
    
    plt.subplot(121)
    plt.hist(data[col], bins=40)
    plt.title(f'{col} - Histogram')

    plt.subplot(122)
    plt.boxplot(data[col], vert=False)
    plt.title(f'{col} - Boxplot')
    
    plt.show()
```

Na osnovu prikazanih histogram i boxplot dijagrama, uočeno je da većina numeričkih atributa u skupu podataka ne prati normalnu raspodelu, već je izražena desno asimetricna raspodela.

Atributi koji se odnose na trajanje poseta pojedinim tipovima stranica tj. Administrative_Duration, Information_Duration, ProductRelated_Duration, pokazuju najizraženiju asimetriju. Većina sesija ima veoma kratko ili nulto trajanje, dok samo mali broj sesija ima znatno duže trajanje. 

Može se zaključiti da prisustvo ekstremnih vrednosti nije posledica grešaka u unosu podataka, već odraz prirodnog ponašanja korisnika na sajtu. Većina sesija podrazumeva kratku i ograničenu interakciju, a značajno manji broj korisnika ostvaruju znatno intenzivniju aktivnost.

```python
for col in num_columns:
    Q1, Q3 = data[col].quantile([0.25, 0.75])
    IQR = Q3 - Q1
    lower, upper = Q1 - 1.5*IQR, Q3 + 1.5*IQR
    outliers = data[(data[col] < lower) | (data[col] > upper)]

    if len(outliers) > 0:
        stopa_kupovine = outliers['Revenue'].mean()
        print(f"{col}: stopa Revenue=True kod outliera = {round(stopa_kupovine * 100,2)}%"
              f"(u odnosu na ukupnu stopu {round(data['Revenue'].mean()*100,2)}%)")

    else:
        print(f'{col}: nema outliera po IQR metodi')
```

Rezultati pokazuju da ekstremne vrednosti kod većine atributa koji opisuju aktivnosti korisnika na sajtu nose značajnu prediktivnu informaciju. Stopa kupovine kod outliera ovih atributa je gotovo duplo veća u odnosu na opštu stopu kupovine. Ovo još jednom ukazuje da sesije sa neuobičajeno visokim brojem poseta stranicama ili neuobičajeno dugim trajanjem interakcije predstavlja korisnike sa izraženijom namerom kupovine, te da uklanjanje ovih vrednosti ne bi bilo opravdano jer bi se izgubila korisna informacija za predviđanje.

# 4.Korelaciona analiza

```python
x = data.drop(['Revenue'], axis=1)
y = data['Revenue']
feature_names = x.columns
```

```python
corrcoefs = []

for feature_name in feature_names:
    corrcoef = np.corrcoef(
        x[feature_name].values, y.values.ravel()
    )[0,1]

    corrcoefs.append(corrcoef)
    print(feature_name, corrcoef)
```

Na osnovu rezultata može se zaključiti da atribut PageValues ima najjaču povezanost sa ciljnom promenljivom. Umerenu pozitivnu korelaciju imaju atributi ProductRelated, ProductRelated_Duration, ali i dummy promenljiva Month_Nov koja verovatno odražava sezonski uticaj praznika i akcija Black Friday-a. 

Sa druge strane promenljiva BounceRates ima umerenu negativnu korelaciju, kao i ExitRates koja pokazuje da sesije sa visokom stopom napuštanja sajta bez dalje interakcije rezultiraju znatno manjom verovatnoćom kupovine. 

```python
corrcoefs = pd.Series(corrcoefs, index = feature_names)
corrcoefs.sort_values(inplace=True)

corrcoefs.plot(kind='barh')
plt.xlabel('Koeficijent korelacije')
plt.title('Korelacija atributa sa ciljnom promenljivom')
plt.show()
```

Kao sto je i prethodno zakljuceno, grafik potvrdjuje da PageValues ima daleko najjacu pozitivnu korelaciju sa Revenue, dok su ExitRates i BounceRates najizraženije negativno korelisani. Ostali atributi pokazuju znatno slabiju linearnu povezanost sa ciljnom promenljivom. 

```python
corr_matrix = x.corr()
corr_matrix
```

```python
plt.figure(figsize=(14,10))

plt.imshow(corr_matrix, cmap='inferno')
plt.colorbar()

plt.xticks(range(len(corr_matrix.columns)), corr_matrix.columns, rotation=90)
plt.yticks(range(len(corr_matrix)), corr_matrix.columns)
plt.title('Matrica korelacije atributa')
plt.show()
```

Matrica korelacije pokazuje da postoji izražena međusobna povezanost između određenih parova atributa. Najjača je između BounceRates i ExitRates (0,91), koje predstavljaju metrike napuštanja sajta. Zatim između ProductRelated i ProductRelated_Duration, Informational i Informational_Duration, kao i Administrative i Administrative_Duration. Ovo je očekivano jer svaki od ova tri para meri srodne veličine, tj. broj poseta stranicama jedne kategorije i vreme provedeno na njima.
SpecialDay i Month_May takođe pokazuju umerenu korelaciju što je i logično s obzirom da Dan majki pada u maju.
Ostali atributi, uključujući i dummy varijable za mesece i tip posetioca, pokazuju slabu međusobnu povezanost.

Heatmapa potvrđuje prethodne zaključke. Jasno se izdvajaju svetlija ostrva van dijagonale koja odgovaraju prethodno pomenutim parovima atributa. Najsvetliji je kvadrat BounceRates i ExitRates, a odmah zatim i blok u gornjem levom uglu koji obuhvata Administrative, Informational i ProductRelated sa svojim Duration parom. Uočava se i izolovana svetlija tačka između SpecialDay i Month_May, dok je ostatak matrice tamniji, što potvrđuje da većina atributa međusobno ne pokazuje značajnu linearnu povezanost.

# 5. Analiza po mesecima          

```python
stopa_po_mesecu = data_original.groupby('Month')['Revenue'].mean().sort_values(ascending=False)

stopa_po_mesecu.plot(kind='bar', figsize=(8,5))

plt.title('Stopa kupovine (Revenue) po mesecima ')
plt.ylabel('Udeo sesija sa Revenue=True')
plt.xlabel('Mesec')

plt.xticks(rotation=90)
plt.show()
```

```python
data_original['Month'].value_counts().sort_values(ascending=False).plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Broj sesija po mesecima')
plt.ylabel('Broj sesija')
plt.xlabel('Mesec')

plt.xticks(rotation=90)
plt.show()
```

Na osnovu prikazanih dijagrama može se primetiti da novembar ima najveću stopu kupovine, iako je broj sesija najveći u maju. To dokazuje da veći broj sesija ne znači nužno i veći broj ostvarenih kupovina.

# 7. Balansiranost klasa 

```python
broj_po_klasi = data['Revenue'].value_counts()
print(broj_po_klasi)
print(data['Revenue'].value_counts())

broj_po_klasi.plot(kind='bar', color=['steelblue', 'orange'], title='Distribucija ciljne promenljive Revenue')
plt.ylabel('Broj sesija')
plt.xticks([0,1], ['Nema kupovinu', 'Kupovina'])
plt.show()
```

Analizom distribucije ciljne promenljive Revenue uočava se izražena neuravnoteženost klasa. Broj sesija kod kojih nije došlo do kupovine značajno je veći od broja sesija koje su rezultovale kupovinom. To ukazuje na to da je većina posetilaca napustila sajt bez izvršene kupovine, dok je samo manji deo sesija završen uspešno, sa ostvarenim prihodom.
